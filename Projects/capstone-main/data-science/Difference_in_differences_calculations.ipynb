<h1> DID Calculations </h1>

```python
import pandas as pd
import json
```

```python
with open("website_data.json") as f:
    data = json.load(f)
```

```python
df = pd.DataFrame(data)
display(df)
```

Georgia and Armenia

```python
# Filtering GNI per capita based on country name and year
filtered = df[
    (df["Country Name"].isin(["Georgia", "Armenia"])) & (df["year"].isin([2015, 2018]))
]

print(filtered[["Country Name", "year", "GNI per Capita"]])
```

```python
# Turning years into colummns
pivoted = filtered.pivot(index="Country Name", columns="year", values="GNI per Capita")
```

```python
# Difference 2018 - 2015
pivoted["GNI_diff_2018_2015"] = pivoted[2018] - pivoted[2015]
result = pivoted["GNI_diff_2018_2015"]
print(result)
```
