# Exploratory Data Analysis (EDA)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# read json
df1 = pd.read_json("website_data.json")
df1["year"] = pd.to_numeric(df1["year"])
df1["Country Name"] = df1["Country Name"].apply(lambda x: str(x).lower())

df2 = pd.read_json("all_data.json")
df2 = df2[["Country Name", "year", "Income group"]]
df2["year"] = pd.to_numeric(df2["year"])
df2["Country Name"] = df2["Country Name"].apply(lambda x: str(x))

df = df1.merge(df2, on=["Country Name", "year"], how="left")
df = df[df["year"] >= 2015]

display(df.head())
```

### 1. Kernal density plot for dataset variance

```python
sns.set_style("darkgrid")

numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns

plt.figure(figsize=(14, len(numerical_columns) * 3))
for idx, feature in enumerate(numerical_columns, 1):
    plt.subplot(len(numerical_columns), 2, idx)
    sns.kdeplot(
        df[feature].dropna(), fill=True
    )  # `fill=True` gives area under the curve
    plt.title(f"{feature} | Skewness: {round(df[feature].skew(), 2)}")

plt.tight_layout()
plt.show()
```

### 2. Swarm Plot for GNI per capita across countries

```python
plt.figure(figsize=(10, 8))

sns.swarmplot(x="GNI per Capita", y="Country Name", data=df, palette="viridis")

plt.title("Swarm Plot for GNI per Capita and country")
plt.xlabel("GNI per Capita")
plt.ylabel("Country")
plt.show()
```

### 3. Swarm Plot for Bandwidth across countries

```python
plt.figure(figsize=(10, 8))

sns.swarmplot(x="bandwidth (kbit/s)", y="Country Name", data=df, palette="viridis")

plt.title("Swarm Plot for Bandwidth and Country")
plt.xlabel("Bandwidth")
plt.ylabel("Country")
plt.show()
```

### 4. Bivarate analysis: pair plot

```python
sns.set_palette("Pastel1")

plt.figure(figsize=(10, 6))

sns.pairplot(df)

plt.suptitle("Pair Plot for DataFrame")
plt.show()
```

### 5. Correlation matrix

```python
# drop nan values
df = df.dropna()
```

```python
# Keep only numeric columns
numeric_df = df.select_dtypes(include="number")
numeric_df = df.drop(columns="year", errors="ignore").select_dtypes(include="number")

plt.figure(figsize=(15, 10))

sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="Pastel2", linewidths=2)

plt.title("Correlation Heatmap")
plt.show()
```

### 6. Kendall's Tau Correlation Heatmap

```python
# Keep only numeric columns
numeric_df = df.select_dtypes(include="number")

# Compute Kendall's Tau correlation matrix
numeric_df = df.drop(columns="year", errors="ignore").select_dtypes(include="number")
kendall_corr = numeric_df.corr(method="kendall")

# Visualize as a heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(
    kendall_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    square=True,
    cbar_kws={"shrink": 0.75},
)
plt.title("Kendall's Tau Correlation Heatmap", fontsize=16)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### 7. Stacked Barcharts showing IP adresses across income groups

```python
# Count total per income group
grouped = df.groupby("Income group")[
    [
        "v4_prefixes_ris",
        "v4_prefixes_stats",
        "v6_prefixes_ris",
        "v6_prefixes_stats",
        "asns_ris",
        "asns_stats",
    ]
].sum()
grouped_sorted = grouped.sort_values("v6_prefixes_ris", ascending=False)

# Get proportions per each income group
grouped_normalised = grouped_sorted.div(grouped_sorted.sum(axis=1), axis=0)

# Visualise
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

grouped_sorted.plot(kind="bar", stacked=True, colormap="Set2", ax=ax[0])
ax[0].set_ylabel("Count")
ax[0].set_title("Count of IP adresses across income groups")

grouped_normalised.plot(kind="bar", stacked=True, colormap="Set2", ax=ax[1])
ax[1].set_ylabel("Proportion")
ax[1].set_title("Proportion of IP adresses across income groups")

plt.show
```

### 8. Boxplots for Speed and Bandwidth across income groups

```python
df = df[df["bandwidth (kbit/s)"] < 8267910]
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
sns.boxplot(x="Income group", y="speed (mgps)", data=df, palette="Set2", ax=ax[0])
ax[0].set_title("Speed (mgps) across income groups")
ax[0].set_xticklabels(ax[0].get_xticklabels(), rotation=10)
sns.boxplot(x="Income group", y="bandwidth (kbit/s)", data=df, palette="Set2", ax=ax[1])
ax[1].set_title("Bandwidth (kbit/s) across income groups")
ax[1].set_xticklabels(ax[1].get_xticklabels(), rotation=10)
plt.show()
```
