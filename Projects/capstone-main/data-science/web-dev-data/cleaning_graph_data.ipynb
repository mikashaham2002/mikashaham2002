## Cleaning and Formatting the Graph Data ##

This data is used to display content on the website, including name and country details, as well as the internet and economic indicators used for visualizations. This notebook also cleans the time series json file used to generate the graphs. The data cleaned includes datasets with actual values and datasets with normalized values which are used for modelling.

```python
# Import libraries

import json
import pandas as pd
```

#### 1. Read and format json file with actual, unscaled values.

```python
# Load json file with all of the data containing the real, unscaled values
file_path = "real_data.json"
with open(file_path, "r", encoding="utf-8") as file:
    raw_data = json.load(file)

real_df = pd.DataFrame(raw_data)
real_df = real_df.rename(columns={"year": "Year"})

# Filter data from 2015
real_df = real_df[(real_df["Year"] >= 2015) & (real_df["Year"] <= 2023)]
display(real_df)
```

```python
# Sort by 'Country Name' and 'Year'
real_df = real_df.sort_values(
    by=["Country Name", "Year"], ascending=[True, True]
).reset_index(drop=True)

# Drop columns not needed for graphs; most of these values are represented through similar metrics
columns = [
    "v4_prefixes_stats",
    "v6_prefixes_stats",
    "asns_stats",
    "GINI Index",
    "GDP per Capita",
    "GDP",
    "GNI",
    "Economic Index",
]
real_df = real_df.drop(columns, axis=1)
display(real_df)
```

```python
# Map names and apply the mapping to standardize country names
name_map = {
    "Bosnia and Herzegowina": "Bosnia and Herzegovina",
    "Bosnia And Herzegovina": "Bosnia and Herzegovina",
    "Iran, Islamic Rep.": "Iran",
    "Kyrgyz Republic": "Kyrgyzstan",
    "Slovak Republic": "Slovakia",
    "West Bank and Gaza": "Palestine",
    "Åland Islands": "Aland Islands",
    "Vatican City State": "Holy See",
    "Türkiye": "Turkey",
    "Turkiye": "Turkey",
    "Yemen, Rep.": "Yemen",
}

real_df["Country Name"] = real_df["Country Name"].str.strip()
real_df["Country Name"] = real_df["Country Name"].replace(name_map)
display(real_df)
```

#### 2. Add internet penetration rate data

```python
from pathlib import Path

# Get path of internet penetration rate csv
BASE_DIR = Path.cwd()
PARENT_DIR = BASE_DIR.parent
file_path = BASE_DIR / "internet_penetration_rate.csv"

ipr_df = pd.read_csv(
    file_path,
    skiprows=4,  # Skip non-data rows
    header=0,
)
# Map names and apply the mapping to standardize country names
ipr_df["Country Name"] = ipr_df["Country Name"].replace(name_map)
display(ipr_df.head())
```

```python
ipr_df = pd.melt(
    ipr_df,
    id_vars=["Country Name"],
    var_name="Year",
    value_name="Internet Penetration Rate",
)

# Convert 'Year' column to numeric values for sorting
ipr_df["Year"] = pd.to_numeric(ipr_df["Year"], errors="coerce")

# Merge dataframes
merged_df = pd.merge(real_df, ipr_df, on=["Country Name", "Year"], how="left")

display(merged_df.head())
```

#### 3. Save in website directory

```python
# Move up two levels to get parent directory
PARENT_DIR = BASE_DIR.parent.parent

# Write path of json file and save to add it in the web development folder where it is extracted for visualizations
file_path = PARENT_DIR / "web-dev" / "app" / "data" / "graph_data.json"
merged_df.to_json(file_path, orient="records", indent=2)
```
