## Collected data used for website content

To provide readable and understanable visualiations or comparisons, the data integrated in the website will be unscaled. The collected data is saved in webstie_data.json and can be found under real_values.json in the web-dev-data folder.

```python
import pandas as pd

# Load data
internet_df = pd.read_json("merged_data.json", lines=True)
economic_df = pd.read_csv("economic_data.csv")
economic_df.rename(columns={"Year": "year"}, inplace=True)

# Merge on 'Country Name' and 'year'
merged_df = pd.merge(
    internet_df,
    economic_df,
    on=["Country Name", "year"],
    how="outer",
    suffixes=("_internet", "_econ"),
)


# Drop any 'Country Code'
for col in ["Country Code", "Country Code_internet", "Country Code_econ"]:
    if col in merged_df.columns:
        merged_df.drop(columns=col, inplace=True)


# Clean column names
def clean_columns(df):
    new_cols = []
    cols_to_keep_idx = []
    for i, col in enumerate(df.columns):
        if col.endswith("_internet"):
            base_col = col.replace("_internet", "")
            if base_col + "_econ" not in df.columns:
                new_cols.append(base_col)
                cols_to_keep_idx.append(i)
            # else: skip internet duplicate
        elif col.endswith("_econ"):
            new_cols.append(col.replace("_econ", ""))
            cols_to_keep_idx.append(i)
        else:
            new_cols.append(col)
            cols_to_keep_idx.append(i)

    df_clean = df.iloc[:, cols_to_keep_idx].copy()
    df_clean.columns = new_cols
    return df_clean


merged_df = clean_columns(merged_df)
display(merged_df)


# Save
merged_df.to_json("website_data.json", orient="records", indent=4)
```
