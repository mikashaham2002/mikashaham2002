## Probabilistic Linkage

Probabilistic linkage, on the other hand, takes into account that data can have errors and inconsistencies. It assigns weights to different matching variables based on their likelihood of being a correct match. Records are linked based on the probability that they match across the datasets, which is determined by a cumulative weight score.

```python
#install recorlinkage package used for illustarting probabilistic matching. Go ahead and use code from the original Data Linkage workshop

!pip install recordlinkage
```

```python
import recordlinkage
import pandas as pd
from recordlinkage.datasets import load_febrl4

# Load a sample datasets used for linkage
df_a, df_b = load_febrl4()

# Convert 'date_of_birth' columns to datetime with a specific format
df_a['date_of_birth'] = pd.to_datetime(df_a['date_of_birth'], format='%Y%m%d', errors='coerce')
df_b['date_of_birth'] = pd.to_datetime(df_b['date_of_birth'], format='%Y%m%d', errors='coerce')
```

Now let's do some record linkage. Here's a breakdown of what each part is doing:

1. **Initialize Record Linkage Indexer:**
   The `indexer` object is created to define how the records between two DataFrames (`df_a` and `df_b`) should be paired for comparison. The `indexer.full()` method specifies that a full index should be built, meaning that every record in `df_a` will be paired with every record in `df_b` to form a Cartesian product, resulting in candidate links.

2. **Initialize Comparison Object:**
   The `compare` object is initialized, which is used to set up different comparison methods for the candidate record pairs.

3. **Add Comparison Logic:**
   Multiple comparison methods are specified for various fields:
   - `compare.exact` is used for an exact match on 'given_name', 'suburb', and 'state'.
   - `compare.string` uses the Jaro-Winkler string comparison method for 'surname' and 'address_1', with a threshold of 0.85. This means records will be considered a match if the comparison score is above this threshold, allowing for some variation in the strings due to typographical errors or different data entry conventions.
   - `compare.date` is set for 'date_of_birth', which will likely compare the dates exactly, assuming they are in the correct datetime format.

4. **Compute Comparison Vectors:**
   The `features` DataFrame is created by computing the comparison vectors for each pair of candidate links across the specified fields.

5. **Classify Matches:**
   Finally, potential matches are classified. In this case, records are considered a match if the sum of their comparison scores across all fields is greater than 3. This threshold is arbitrary and would typically be determined based on the nature of the data and the desired balance between match precision and recall.

```python
# Initialize the record linkage object
indexer = recordlinkage.Index()
indexer.full()
candidate_links = indexer.index(df_a, df_b)

# Initialize the comparison object
compare = recordlinkage.Compare()

# Add comparison logic for different fields
compare.exact('given_name', 'given_name', label='given_name')
compare.string('surname', 'surname', method='jarowinkler', threshold=0.85, label='surname')
compare.date('date_of_birth', 'date_of_birth', label='date_of_birth')
compare.exact('suburb', 'suburb', label='suburb')
compare.exact('state', 'state', label='state')
compare.string('address_1', 'address_1', method='jarowinkler', threshold=0.85, label='address_1')

# The comparison vectors
features = compare.compute(candidate_links, df_a, df_b)

# Classify the matches
matches = features[features.sum(axis=1) > 3]
```
