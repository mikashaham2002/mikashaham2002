Cleaning DNNL Data

```python
import pandas as pd
import string

#Read and clean the database
excel_file_path = 'DNNL DATA.xlsx'
columns_to_use = ['First Name', 'Last Name', 'Company Name']
df = pd.read_excel(excel_file_path, usecols=columns_to_use)
df = df.dropna()

#Create column with full names
df['Full Name']=df['First Name'] + ' ' + df['Last Name']
display(df)

#Convert full name column to list
full_names=df['Full Name'].tolist()

#Convert elements in list to strings and remove 
#unwanted words or signs to further clean data
full_names = [str(name) for name in full_names]
remove_words = "(esdp-ic)"
full_names = [word.replace(remove_words, '') for word in full_names]

print(full_names)
```

Scrape Linekdin Profile Picture

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

#open LinkedIn website
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))

time.sleep(5)
driver.get("https://linkedin.com/uas/login")
time.sleep(5)
```

```python
#Log in
# user = "maria.baba1886@gmail.com"
# passw = "DNNLTestScraper18@"
user = "rosasalias20@gmail.com"
passw = "DNNLtest40"

username=driver.find_element(By.ID, 'username')
username.send_keys(user)
time.sleep(4)
password=driver.find_element(By.ID, 'password')
password.send_keys(passw)
time.sleep(3)
driver.find_element(By.XPATH, "//button[@type='submit']").click()
```

```python
#Get profile pictures


#Create empty list to collect all image URLs
all_srcs=[]

#For loop to iterate through each person
for person in full_names:
    try:
#Click on the search bar and type name
        search_bar=driver.find_element(By.CLASS_NAME, 'search-global-typeahead__input')
        search_bar.click()
        time.sleep(3)
        search_bar.send_keys(person)
        time.sleep(2)
        search_bar.send_keys(Keys.ENTER)
        time.sleep(8)

#Get profile picture
        image = driver.find_element(By.CSS_SELECTOR, 'img[src*="profile-displayphoto-shrink"]')
        image_src = image.get_attribute("src")
        time.sleep(6)

#Clear search bar so next name can be written
        search_bar.clear()

#Append image source to URL list
        all_srcs.append(image_src)

#handle exceptions as undefined sources
    except Exception as e:
        all_srcs.append('undefined')
        search_bar.clear()
#Print image sources
print(all_srcs)
```

```python
#Get about sections

#Create empty list to collect about sections

all_abouts = []

#For loop to iterate through each person
for person in full_names:
#Click on the search bar and type name
    search_bar=driver.find_element(By.CLASS_NAME, 'search-global-typeahead__input')
    search_bar.click()
    time.sleep(3)
    search_bar.send_keys(person)
    time.sleep(1)
    search_bar.send_keys(Keys.ENTER)
    time.sleep(4)

#Get profile picture
    #image = driver.find_element(By.CSS_SELECTOR, 'img[src*="profile-displayphoto-shrink"]')
    #image_src = image.get_attribute("src")
    #time.sleep(1)
# Click on result 
    #image.click()
    driver.execute_script("window.scrollBy(0, 700);")

# Try to click more text if available
    try:
        see_more = driver.find_element(By.XPATH, '//button[contains(., "see more")]')

        see_more.click()
    except:
        None

    # Locate the about section
    about_section = driver.find_element(By.XPATH, '//div[contains(@class, "ph5") and contains(@class, "pv3")]')

    # Get the text content of the about section
    about_text = about_section.text
    time.sleep(3)

#Clear search bar so next name can be written
    search_bar.clear()

#Append image source to URL list + about section
    #all_srcs.append(image_src)
    all_abouts.append(about_text)
#Print image sources + abouts
print(all_abouts)
```

```python
import csv
#Save image sources to csv file
csv_name = 'Images_URL.csv'

# Write to the CSV file
with open(csv_name, 'w', newline='') as csv_file:
    csv_writer = csv.writer(csv_file)
    for x in all_srcs:
        csv_writer.writerow([x])

```

```python
#Save about sections to csv file
csv_name = 'about_texts.csv'

# Write to the CSV file
with open(csv_name, 'w', newline='') as csv_file:
    csv_writer = csv.writer(csv_file)
    for x in all_abouts:
        csv_writer.writerow([x])
```
