Cleaning DNNL Data

```python
import pandas as pd
import string

#Read and clean the database
excel_file_path = 'DNNL DATA.xlsx'
columns_to_use = ['First Name', 'Last Name', 'Company Name']
df = pd.read_excel(excel_file_path, usecols=columns_to_use)
df = df.dropna()
df = df.reset_index(drop=True)

#Create column with full names
df['Full Name']=df['First Name'] + ' ' + df['Last Name']
display(df)

#Convert full name column to list
full_names=df['Full Name'].tolist()

#Convert elements in list to strings and remove 
#unwanted words or signs to further clean data
full_names = [str(name) for name in full_names]
remove_words = "(esdp-ic)"
full_names = [word.replace(remove_words, '') for word in full_names]

print(full_names)
```

```python
#Create empty list to collect about sections

all_abouts = []

#For loop to iterate through each person
for person in full_names:
    try:
#Click on the search bar and type name
        search_bar=driver.find_element(By.CLASS_NAME, 'search-global-typeahead__input')
        search_bar.click()
        time.sleep(3)
        search_bar.send_keys(person)
        time.sleep(1)
        search_bar.send_keys(Keys.ENTER)
        time.sleep(4)

#Get profile picture
        image = driver.find_element(By.CSS_SELECTOR, 'img[src*="profile-displayphoto-shrink"]')
        image_src = image.get_attribute("src")
        time.sleep(1)
# Click on result 
        image.click()
        driver.execute_script("window.scrollBy(0, 700);")

# Locate the about section
        time.sleep(4)
        about_section = driver.find_element(By.XPATH, '//div[contains(@class, "ph5") and contains(@class, "pv3")]')
        time.sleep(3)
# Get the text content of the about section
        about_text = about_section.text
        time.sleep(4)

#Clear search bar so next name can be written
        search_bar.clear()
        all_abouts.append(about_text)

    except Exception as e:
        all_abouts.append('None')
        search_bar.clear()
#Print image sources + abouts
print(all_abouts)
```

```python
# Create a new column to store the matched goals
merged_df['Matched_SDGs'] = ""

# Iterate through each row and check for keyword matches
for index, row in merged_df.iterrows():
    about_text = row['About']
    if isinstance(about_text, str):
        matched_goals = [goal for goal, keywords in sdg_keywords.items() if any(keyword in about_text for keyword in keywords)]
    else:
        matched_goals = 'No goals match'

    # Update the 'Matched_Goals' column with the matched goals
    merged_df.at[index, 'Matched_SDGs'] = ', '.join(matched_goals)

merged_df.dropna(subset=['About'], inplace=True)

# Display the updated DataFrame
display(merged_df)
```

```python
import pandas as pd
merged_df = pd.read_excel('LinkedIn/Sustainable_development_goals_entrepreneurs.xlsx')
display(merged_df)
merged_df['Matched_SDGs'] = merged_df['Matched_SDGs'].astype(str)

#View the SDGs found in the database
print(merged_df['Matched_SDGs'].unique())

#Split SDG column so each goal is on its own column in a new dataframe
merged_df = merged_df['Matched_SDGs'].str.split(',|\n', expand=True)
SDG = merged_df.stack().reset_index(level=1, drop=True).to_frame('Column')

# Remove whitespaces
SDG['Column'] = SDG['Column'].str.strip()

# Calculate total of entrepreneurs in the dataframe
total= len(SDG)

# Calculate percentages of entrepreneurs involved in the SDGs
thirteen = (SDG['Column'] == 'SDG13').sum()
f13= ((thirteen)/(total))*100
p13= round(f13)

sixteen = (SDG['Column'] == 'SDG16').sum()
f16= ((sixteen)/(total))*100
p16= round(f16)

three = (SDG['Column'] == 'SDG3').sum()
f3= ((three)/(total))*100
p3= round(f3)

two = (SDG['Column'] == 'SDG2').sum()
f2= ((two)/(total))*100
p2= round(f2)

six = (SDG['Column'] == 'SDG6').sum()
f6= ((six)/(total))*100
p6= round(f6)

five = (SDG['Column'] == 'SDG5').sum()
f5= ((five)/(total))*100
p5= round(f5)

eleven = (SDG['Column'] == 'SDG11').sum()
f11= ((eleven)/(total))*100
p11= round(f11)

twelve = (SDG['Column'] == 'SDG12').sum()
f12= ((twelve)/(total))*100
p12= round(f12)

fifteen = (SDG['Column'] == 'SDG15').sum()
f15= ((fifteen)/(total))*100
p15= round(f15)
```
