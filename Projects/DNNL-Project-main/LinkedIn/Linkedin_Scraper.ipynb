```python
import pandas as pd

df = pd.read_excel("DNNL DATA.xlsx")

df.dropna(subset=['Associated Company'], inplace=True)

associated_companies = df['Associated Company'].unique().tolist()

print(associated_companies)
```

```python
industries = {}
for company_name in associated_companies[:1]:
    search = driver.find_element("class name", "search-global-typeahead__input")
    search.clear()
    search.click()
    search.send_keys(company_name)
    search.send_keys(Keys.ENTER)
    time.sleep(1.5)
    company_link = driver.find_element(By.XPATH, "//span[contains(text(), 'View page')]")
    company_link.click()

    time.sleep(1)
    about_text = driver.find_elements(By.CLASS_NAME, 'mb4.t-black--light.text-body-medium')
    company_industry = about_text[1].text
    print(company_industry)
    industries[company_name] = company_industry
```
