```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#Read csv file with leapfunder data
columns= ['name','position','bio','Gender','Female Representation %','Male Representation %']
df= pd.read_csv('gender_data_leapfunder.csv', names=columns)

#Drop unecessary columns to clean dataframe
columns_to_drop = ['name', 'bio', 'Female Representation %', 'Male Representation %']
df.drop(columns=columns_to_drop, inplace=True)
display(df)
```

```python

# Create a function to group positions based on the same meaning or synonyms
def group_positions(position):
    if pd.notna(position): 
        keywords = {
            'CEO': ['Chief Executive Officer', 'CEO', 'Leader'],
            'Specialist in a field': ['AI/ML', 'Data Science','Technology', 'Back-end', 'Front-end','Designer','Branding' ],
            'Founder': ['Founder', 'Co-founder', 'Owner'],
            'Chief': ['CPO', 'CTO', 'COO', 'Chief', 'CFO'],
            'Director': ['General Director', 'Sales director', 'Director', 'Directie'], 
            'Advisor': ['Advisor', 'advisor'],
            'Partner':['Partner'], 
            'Head': ['Head'],
            'Manager': ['Manager'],
            'Consultant': ['Consultant'],
            'Board Member': ['Board'],
            'Investor': ['Investor']
        }
        for keyword, synonyms in keywords.items():
            for synonym in synonyms:
                if synonym in position:
                    return keyword
    return 'Other'

#Apply function for values in positon column
df['Cleaned_Position'] = df['position'].apply(group_positions)

#Remove values named as 'other'. This values are deemed insignificant for the purpose of the visualization.
df = df[df['Cleaned_Position'] != 'Other']

#Get proportions of gender by position
position_gender=df.groupby(['Cleaned_Position', 'Gender']).size()
position=df.groupby('Cleaned_Position').size()

proportions= position_gender/ position

#Add proportion to dataframe
column= 'Proportion'
proportions = proportions.reset_index(name=column)


# Create a bar plot for gender and cleaned position with proportions
plt.figure(figsize=(12, 8))
sns.barplot(x='Cleaned_Position', y='Proportion', hue='Gender', data=proportions, palette='pastel')
plt.title('Gender Distribution by Position')
plt.xlabel('Position in Company')
plt.ylabel('Proportion of gender per position')
plt.xticks(rotation=45, ha='right')
plt.show()
```
