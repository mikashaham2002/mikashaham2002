```python
import json
import requests
#Create function for nationality API
def identify_nationality(first_name):
    base= 'https://api.nationalize.io?'

    url= f'{base}name={first_name}'
    result=requests.get(url)
    
    json_str = result.text
    details= json.loads(json_str)
    return details
# Test
res= identify_nationality("Maria")
print(res)

countries= res['country'][0]

country_id= countries["country_id"]
print(country_id)
```

```python
import pandas as pd
import numpy as np
#Read csv file and extract first names
data= pd.read_csv("gender_data_leapfunder.csv")
name= data['name']
first_name= [x.split()[0] for x in name]


#Apply nationality function for first names
countries=[]
for x in first_name:
    value= identify_nationality(x)
    countries.append(value)
```

```python
#Store the dictionaries in a pickle for easy access.
import pickle
output = "countries.pkl"
with open(output, "wb") as f:
    pickle.dump(countries, f)
```

```python
df_countries = pd.read_pickle('countries.pkl')
```

```python
#Get count for name and eliminate all values that have 0
#These values do not have a country_id to show nationality
lst=[]
for x in df_countries:
    #print(x)
    count= x['count']
    if count>0:
        lst.append(x)
```

```python
#Get the country_id (nationality) based on the highest probability (first country).
#Append IDs to list.
list_of_countries=[]
for value in lst:
    country= value['country'][0]
    #print(country)
    nationality= country['country_id']
    print(nationality)
    list_of_countries.append(nationality)
```

```python
#Get the names of founders from the Nationalize API dictionary created
name=[]
for value in lst:
    names=value['name']
    name.append(names)
```

```python
#Create dataframe with nationality and first name.
df = pd.DataFrame(list(zip(list_of_countries, name)), columns =['nationality', 'first'])
```

```python
#Extract first name values from the orginal csv file and add to dataframe.
#By extracting this, we can later merge the two dataframes 
#by first names to add nationality
data= pd.read_csv("gender_data.csv")
values= data['name']

first_names_all=[]
for x in values:
    first=x.split()[0]
    first_names_all.append(first)

data['first']=first_names_all
```

```python
#Merge dataframes on first names.
#Save results in a new csv file for nationalities
result = pd.merge(data, df, on=['first'])
result = result.drop_duplicates()


result.to_csv('Nation_Leapfunder.csv', index=False)
```
