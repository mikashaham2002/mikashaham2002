```python
import pandas as pd
from tabulate import tabulate


#split og data by ;
df2 = df['PROFIT'].str.split(';', expand=True)

#replace n/a values
df2 = df2.apply(lambda x: x.str.replace('n/a', '0')).apply(pd.to_numeric)

#find mean
avg = df2.mean(axis=1)
df2['MEAN'] = avg

#replace NaN values with fillna
df2 = df2.fillna(0)

#rename columns for clarity
df2.columns = ['2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', 'MEAN']

#make it look good using tabulate
table= tabulate(df2, headers='keys', tablefmt='pretty', showindex=False)

print(table)

```
