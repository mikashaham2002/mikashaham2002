```python
#Create a stratup list containing lists of dictionaries with links of the company and the names

startups = []


#Find link elements of each startup and save it in a list
link_elements = driver.find_elements(By.XPATH, "//a[contains(@class, 'entity-name__name-text')]")

#Access each link element in the link_elements list
for link_element in link_elements:
    #Get the links and names of companies, save it as a dictionary in a startups list
    try:
        if link_element: 
            url = link_element.get_attribute('href')
            company_name = link_element.text
            startups.append({'Link':url,'Company name': company_name })

    except StaleElementReferenceException:
        pass


#Convert startup list to Dataframe and delete duplicates
startups = pd.DataFrame(startups)
startups = startups.drop_duplicates()

startups_amount = len(startups['Link'])
print(f'You successfully collected {startups_amount} links')
```

```python
#List with industry categories
industry_list = ['health', 'fintech', 'food', 'media', 'marketing', 'fashion', 'enterprise software',
                 'transportation', 'home living', 'wellness beauty', 'real estate', 'kids', 'education', 'energy', 'sports',
                 'gaming', 'hosting', 'travel', 'semiconductors', 'music', 'event tech', 'jobs recruitment', 'security',
                 'legal', 'robotics', 'telecom', 'dating', 'service provider', 'engineering and manufacturing equipment',
                 'space', 'consumer electronics', 'chemicals']

#Save links and company names in variables
links = startups["Link"]
company_names = startups['Company name']

#Create a list company_data with link, and name of each company, industry, and the list 'founders'
company_data = []

#Iterate through links and names at the same time
for startup_link, company_name in zip(links, company_names):
    company_name = company_name
    #Access the link
    driver.get(startup_link)
    
    time.sleep(2) 

    #Collect industry information about each company
    try:
        industries = []
        #Find industry elements
        industry_elements =  driver.find_elements(By.CLASS_NAME, 'tag-item__name')
        for industry_element in industry_elements:
            #Get industry name
            industry = industry_element.text
            #Check if the name is in list
            if industry.lower() in industry_list:
                industries.append(industry)
        #If list is empty, industry uknown
        if not industries:
            industries = 'Unknown'
    except:
        pass

    #Collect funding information about each company
    try:
        funding = []
        #Find funding elements
        funding_elements =  driver.find_elements(By.CLASS_NAME, 'tooltip-target')
        #Get funding info and put it in a list
        for funding_element in funding_elements:
            funding_amount = funding_element.text
            funding.append(funding_amount)
        #If list is empty, funding unknown
        if not funding or '-' in funding:
            funding = 'Unknown'
    except: 
        pass
    #Access 'people' tab 
    try:
        founders_button = driver.find_element(By.CSS_SELECTOR, "#window-scrollbar > div:nth-child(1) > main > div > div > div.tab-links.entity-profile-tabs.entity-profile-tabs--tab-buttons > button:nth-child(4)")
        founders_button.click()
    except:
        continue
    
    time.sleep(3)
    
    #Create founders list with names and titles of each founder in each company
    founders = []
    #Get elements of names and titles of people in 'people' tab
    founders_names = driver.find_elements(By.XPATH, "//a[@class='entity-name__name-text']")
    founders_titles = driver.find_elements(By.CLASS_NAME, 'user-positions__title')

    #Iterate through name and title elements 
    for name, title in zip(founders_names, founders_titles):
        #Access the text in those elements and save them in a dictionary
        founder_data = {'name': name.text, 'title': title.text}
        if founder_data['title'] == '':
            founder_data['title'] = 'Unknown'
        #Append the founders list
        founders.append(founder_data)
    #Append the company_data list
    company_data.append({'Link': startup_link, 'Company name': company_name, 'Industries': industries, 'Funding': funding, 'Founders data': founders})

#Save company_data list as dataframe and csv
company_data = pd.DataFrame(company_data)
company_data.to_csv('dealroom_data_by_company_new.csv', index=False)
```

```python
#Converting 'dealroom_data_by_company.csv' to a 'dealroom_data_by_person.csv'
#Converting the structure of the csv, each row will now be data of the person, not company

#Create founders_data list
data = pd.read_csv('dealroom_data_by_company.csv')
founders_data_by_people = []

#Iterate through the company_data DataFrame
for index, row in data.iterrows():
    founders_data = eval(row['Founders data'])
    
    #Iterate through list of dictionaries saved in a column 'Founders Data'
    for founder in founders_data:
        name = founder.get('name', 'Unknown')    
        title = founder.get('title', 'Unknown')
        founders_data_by_people.append([name, title, row['Company name'], row['Link']])

#Create a new DataFrame with the expanded founder data
founders_df = pd.DataFrame(founders_data_by_people, columns=['Name', 'Title', 'Company name', 'Link'])

#Save as csv
founders_df.to_csv('dealroom_data_by_person.csv_new', index=False)
```

```python
#Gender API
def identify_gender(first_name):
    base= 'https://api.genderize.io?'

    url= f'{base}name={first_name}'
    result=requests.get(url)
    
    json_str = result.text
    details= json.loads(json_str)
    return details
```

```python
import numpy as np
data = pd.read_csv("dealroom_data_by_person_new.csv", header=0)
name= data['Name']
first_name = [x.split()[0] if isinstance(x, str) and len(x.split()) >= 2 else np.NaN for x in name]
gender= [identify_gender(x)for x in first_name]

estimated_gender= [x['gender'] for x in gender]
df = data.assign(Gender= estimated_gender)
    
#Overwrite csv file
df.to_csv('dealroom_data_by_person_new.csv', index=False)
```
